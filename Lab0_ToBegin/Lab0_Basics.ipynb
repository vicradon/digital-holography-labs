The basics for loading an image in Python

```python
img=cv.imread('Images/cameraman.png', cv.IMREAD_GRAYSCALE)
```

#### How to acces to the gray level of an image in Python

```python
img[1,2] #numpy matrices are ecoded as such: img[row,column]. img[1,2] gives the gray level of the second row and 3rd column 
```

#### Basics for displaying an image in Python

```python
plt.figure(),
plt.subplot(1,2,1),
plt.imshow(img,'gray'),
plt.colorbar()
plt.subplot(1,2,2),
plt.imshow(img-50,'gray'),
plt.colorbar(),
plt.show()
```

#### Loop in Python

```python
for c in range(10):
    c=c+1
    c=c*c
    print(c, end=" ")    
```

Looping through a list:

```python
li=[1, 5, 7, 3, 8, 9, 4, 6]
for i,l in enumerate(li): # i is the index of the element l
    print(f"({i},{l})", end=" ") # f"" are formatted strings that allow parsing variables by writing {varaible_name}
```
