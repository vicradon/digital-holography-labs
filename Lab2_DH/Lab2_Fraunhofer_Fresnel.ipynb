### 5. Simulation of the diffraction pattern recorded on a camera (of size 1024x1024 pixels with pixel pitch $pp=10 \mu m$) and display of the axis using metric units (horizontal and vertical coordinates)

```python
def J1c(x):
    index_0 = np.where(x == 0)
    x[index_0[0],index_0[1]]=1
    y=sp.special.jv(1,x)/x
    y[index_0[0],index_0[1]]=0.5
    return y
```

```python
# lines below should be completed using information given in the text of the lab (Lab2_DH.pdf) and turn from "raw" to "code"
W=1024
H=1024
pp=10e-6
a=150e-6 # a is the radius of the desitination
z=1.5
wavelength=0.6e-6

y=np.arange(-W/2,W/2-1,1)
y=y*pp

x=np.arange(-H/2,H/2-1,1)
x=x*pp

[X,Y]=np.meshgrid(x,y)

arg=2*np.pi*a*np.sqrt(X**2+Y**2)/wavelength/z
U_Frau=(np.pi*a**2)*2*J1c(arg) 
I_Frau=1/(wavelength * z)**2 * U_Frau**2

plt.figure(layout='compressed')
plt.subplot(1,2,1)
plt.imshow(I_Frau,'gray',extent=[(-W*pp/2)*1000,(W*pp/2)*1000,(-H*pp/2)*1000,(H*pp/2)*1000])
plt.colorbar()
plt.xlabel('x(mm)')
plt.ylabel('y(mm)')
plt.title('Intensity Fraunhofer \n Analytical expression')

plt.subplot(1,2,2)
plt.plot(x,I_Frau[512,:])
plt.xlabel('x(mm)')
plt.ylabel('y(mm)')
plt.title('Profile Intensity Fraunhofer \n Analytical expression')
plt.show()
```
